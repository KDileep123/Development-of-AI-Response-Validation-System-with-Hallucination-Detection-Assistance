# SQL Queries Practice

**Objective:** Practice common SQL queries using a sample student database.

**Topics:** `CREATE TABLE`, `INSERT`, `SELECT`, `WHERE`, `ORDER BY`, `DISTINCT`, aggregate functions, `GROUP BY`, `HAVING`, `UPDATE`, `DELETE`, `LIKE`, `BETWEEN`, subqueries, and joins.

These examples use SQLite syntax and can be run in Google Colab. The setup cell creates a temporary database and sample records.

```python
import sqlite3
import pandas as pd

connection = sqlite3.connect(":memory:")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE students (
    Student_ID INTEGER PRIMARY KEY,
    Name TEXT NOT NULL,
    Department TEXT,
    Attendance INTEGER,
    Final_Marks INTEGER
)
""")

students = [
    (101, "Kavya", "BCA", 88, 86),
    (102, "Arjun", "BCS", 79, 77),
    (103, "Ishita", "BCA", 94, 93),
    (104, "Manoj", "BCS", 67, 65),
    (105, "Divya", "BCA", 83, 82),
    (106, "Kiran", "BCS", 72, 70),
    (107, "Nisha", "BCS", 96, 95),
    (108, "Varun", "BCA", 74, 73),
    (109, "Meera", "BCS", 86, 85),
    (110, "Sanjay", "BCA", 61, 60)
]
cursor.executemany("INSERT INTO students VALUES (?, ?, ?, ?, ?)", students)
connection.commit()

print("Student table created and sample records inserted.")
```

## Query 1: Display all records
Use `SELECT *` to display every column and row.

```python
SELECT_ALL = "SELECT * FROM students ORDER BY Student_ID;"
pd.read_sql_query(SELECT_ALL, connection)
```

## Query 2: Select specific columns
Display each student's name, department, and final marks.

```python
query = """
SELECT Name, Department, Final_Marks
FROM students
ORDER BY Name;
"""
pd.read_sql_query(query, connection)
```

## Query 3: Filter with WHERE
Find students whose final marks are greater than 80.

```python
query = """
SELECT Name, Final_Marks
FROM students
WHERE Final_Marks > 80
ORDER BY Final_Marks DESC;
"""
pd.read_sql_query(query, connection)
```

## Query 4: Sort with ORDER BY
List students from highest to lowest final marks.

```python
query = """
SELECT Name, Final_Marks
FROM students
ORDER BY Final_Marks DESC;
"""
pd.read_sql_query(query, connection)
```

## Query 5: DISTINCT values
Display each department only once.

```python
query = "SELECT DISTINCT Department FROM students ORDER BY Department;"
pd.read_sql_query(query, connection)
```

## Query 6: Aggregate functions
Calculate the number of students, average marks, highest marks, and lowest marks.

```python
query = """
SELECT COUNT(*) AS Student_Count,
       ROUND(AVG(Final_Marks), 2) AS Average_Marks,
       MAX(Final_Marks) AS Highest_Marks,
       MIN(Final_Marks) AS Lowest_Marks
FROM students;
"""
pd.read_sql_query(query, connection)
```

## Query 7: GROUP BY
Calculate the average final marks for each department.

```python
query = """
SELECT Department, ROUND(AVG(Final_Marks), 2) AS Average_Marks
FROM students
GROUP BY Department
ORDER BY Department;
"""
pd.read_sql_query(query, connection)
```

## Query 8: HAVING
Show departments whose average final marks are above 75.

```python
query = """
SELECT Department, ROUND(AVG(Final_Marks), 2) AS Average_Marks
FROM students
GROUP BY Department
HAVING AVG(Final_Marks) > 75
ORDER BY Department;
"""
pd.read_sql_query(query, connection)
```

## Query 9: BETWEEN
Find students whose final marks are between 70 and 85, inclusive.

```python
query = """
SELECT Name, Final_Marks
FROM students
WHERE Final_Marks BETWEEN 70 AND 85
ORDER BY Final_Marks;
"""
pd.read_sql_query(query, connection)
```

## Query 10: LIKE
Find students whose names start with the letter 'M'.

```python
query = """
SELECT Name
FROM students
WHERE Name LIKE 'M%'
ORDER BY Name;
"""
pd.read_sql_query(query, connection)
```

## Query 11: Subquery
Find students who scored above the overall average final marks.

```python
query = """
SELECT Name, Final_Marks
FROM students
WHERE Final_Marks > (SELECT AVG(Final_Marks) FROM students)
ORDER BY Final_Marks DESC;
"""
pd.read_sql_query(query, connection)
```

## Query 12: INNER JOIN
Create a departments table and show each student with the department head.

```python
cursor.execute("""
CREATE TABLE departments (
    Department TEXT PRIMARY KEY,
    HOD TEXT
)
""")
cursor.executemany(
    "INSERT INTO departments (Department, HOD) VALUES (?, ?)",
    [("BCA", "Dr. Sharma"), ("BCS", "Dr. Mehta")]
)
connection.commit()

query = """
SELECT students.Name, students.Department, departments.HOD
FROM students
INNER JOIN departments
ON students.Department = departments.Department
ORDER BY students.Student_ID;
"""
pd.read_sql_query(query, connection)
```

## Query 13: LEFT JOIN
A `LEFT JOIN` keeps all rows from the left table, even if no matching department exists.

```python
query = """
SELECT students.Name, students.Department, departments.HOD
FROM students
LEFT JOIN departments
ON students.Department = departments.Department
ORDER BY students.Student_ID;
"""
pd.read_sql_query(query, connection)
```

## Query 14: UPDATE
Update one student's attendance, then display the changed value.

```python
cursor.execute("UPDATE students SET Attendance = 90 WHERE Student_ID = 102")
connection.commit()
pd.read_sql_query("SELECT Student_ID, Name, Attendance FROM students WHERE Student_ID = 102;", connection)
```

## Query 15: DELETE
Delete a temporary sample record and verify it was removed.

```python
cursor.execute("INSERT INTO students VALUES (999, 'Temporary', 'BCA', 50, 50)")
cursor.execute("DELETE FROM students WHERE Student_ID = 999")
connection.commit()
print("Temporary record count:", cursor.execute("SELECT COUNT(*) FROM students WHERE Student_ID = 999").fetchone()[0])
```

## Quick revision

- `SELECT` retrieves data.
- `WHERE` filters rows.
- `ORDER BY` sorts results.
- `GROUP BY` groups rows for aggregate calculations.
- `HAVING` filters grouped results.
- `JOIN` combines related tables.
- `UPDATE` modifies rows; `DELETE` removes rows.
