## 3. What are Python data types?

Common built-in types include `int`, `float`, `str`, `bool`, `list`, `tuple`, `set`, `dict`, and `NoneType`.

## 4. List vs Tuple

A list is mutable, while a tuple is immutable.

```python
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)

my_list[0] = 100          # lists can be changed
print("List after change:", my_list)

try:
    my_tuple[0] = 100     # tuples cannot
except TypeError as e:
    print("Tuple error:", e)
```

## 5. List vs Set

A list is ordered and allows duplicates. A set stores unique elements and does not support indexing like a list.

```python
my_list = [1, 2, 2, 3, 3]
my_set = set(my_list)

print("List:", my_list, "(keeps order and duplicates)")
print("Set :", my_set, "(unique only)")
print("list[0] =", my_list[0])
try:
    my_set[0]
except TypeError as e:
    print("Set indexing error:", e)
```

## 6. What is a dictionary?

A dictionary stores data as key-value pairs.

```python
student = {"name": "Asha", "age": 20, "branch": "CSE"}

print(student["name"])
student["age"] = 21          # update
student["city"] = "Hyderabad"  # add
print(student)
for key, value in student.items():
    print(key, "->", value)
```

## 7. What is mutable and immutable?

Mutable objects can be changed after creation. Immutable objects cannot be changed after creation.

```python
# Mutable: list
a = [1, 2, 3]
print("id before:", id(a))
a.append(4)
print("id after :", id(a), "-> same object, content changed:", a)

# Immutable: string
s = "hello"
print("\nid before:", id(s))
s = s + " world"
print("id after :", id(s), "-> new object created:", s)
```

## 8. Difference between == and is

`==` compares values, while `is` checks whether two references point to the same object.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b :", a == b, "(same values)")
print("a is b :", a is b, "(different objects)")
print("a is c :", a is c, "(same object)")
```

## 9. What is a function?

A function is a reusable block of code that performs a particular task.

```python
def add(x, y):
    """Return the sum of two numbers."""
    return x + y

print(add(5, 3))
print(add(10, 20))
```

## 10. What is exception handling?

Exception handling manages runtime errors using constructs such as `try`, `except`, `else`, and `finally`.

```python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: cannot divide by zero")
    except TypeError:
        print("Error: inputs must be numbers")
    else:
        print("Result:", result)
    finally:
        print("Done (finally always runs)\n")

divide(10, 2)
divide(10, 0)
divide(10, "a")
```

## 11. What is OOP?

Object-Oriented Programming organizes programs using classes and objects.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def show(self):
        print(f"{self.name} scored {self.marks}")

s1 = Student("Asha", 90)
s2 = Student("Ravi", 85)
s1.show()
s2.show()
```

## 12. What are the main OOP concepts?

Encapsulation, inheritance, polymorphism, and abstraction.

```python
from abc import ABC, abstractmethod

# Encapsulation: data hidden behind methods
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance          # private attribute
    def deposit(self, amount):
        self.__balance += amount
    def get_balance(self):
        return self.__balance

acc = BankAccount(1000)
acc.deposit(500)
print("Encapsulation  - balance:", acc.get_balance())

# Abstraction: hide details, force subclasses to implement
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side ** 2

print("Abstraction    - square area:", Square(4).area())
print("(Inheritance and polymorphism are shown in the next two questions)")
```

## 13. What is inheritance?

Inheritance allows a class to acquire properties and methods from another class.

```python
class Animal:
    def speak(self):
        print("Some sound")

class Dog(Animal):          # Dog inherits from Animal
    def bark(self):
        print("Woof!")

d = Dog()
d.speak()   # inherited from Animal
d.bark()    # its own method
```

## 14. What is polymorphism?

Polymorphism means the same interface or method name can have different behavior depending on the object.

```python
class Dog:
    def speak(self):
        return "Woof"

class Cat:
    def speak(self):
        return "Meow"

for animal in [Dog(), Cat()]:
    print(animal.speak())      # same method name, different behaviour
```

## 15. What is list comprehension?

List comprehension is a concise way to create a list from an iterable.

```python
squares = [x * x for x in range(5)]
print("Squares:", squares)

evens = [x for x in range(10) if x % 2 == 0]
print("Evens  :", evens)
```

## 16. What are *args and **kwargs?

`*args` accepts a variable number of positional arguments, while `**kwargs` accepts a variable number of keyword arguments.

```python
def demo(*args, **kwargs):
    print("args  :", args)
    print("kwargs:", kwargs)

demo(1, 2, 3, name="Asha", age=20)

def total(*numbers):
    return sum(numbers)

print("Sum:", total(1, 2, 3, 4, 5))
```

## 17. What is a lambda function?

A lambda is a small anonymous function written using the `lambda` keyword.

```python
square = lambda x: x * x
add = lambda a, b: a + b
print(square(5))
print(add(3, 4))

# Common use: sorting with a key
students = [("Asha", 90), ("Ravi", 75), ("Priya", 85)]
print(sorted(students, key=lambda s: s[1]))
```

## 18. What is a module?

A module is a Python file containing reusable code such as functions, classes, or variables.

## 19. What is file handling?

File handling allows programs to read from and write to files.

```python
# Write
with open("sample.txt", "w") as f:
    f.write("Hello, Python!\n")
    f.write("File handling example.\n")

# Read
with open("sample.txt", "r") as f:
    print(f.read())

# Append
with open("sample.txt", "a") as f:
    f.write("Appended line.\n")

with open("sample.txt") as f:
    print("After append:")
    print(f.read())
```

## 20. What is the difference between shallow copy and deep copy?

A shallow copy creates a new outer object but may share nested objects. A deep copy recursively copies nested objects as well.

```python
import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)
deep = copy.deepcopy(original)

original[0][0] = 99    # change a nested element

print("Original:", original)
print("Shallow :", shallow, "-> affected (shares nested lists)")
print("Deep    :", deep, "-> not affected")
```

# Part 2: SQL Interview Questions

The examples use a small sample database created with Python's built-in `sqlite3` library. Run the setup cell first.

## SQL 1. What is SQL?

SQL (Structured Query Language) is used to create, read, update, and delete data in relational databases. Common commands include `SELECT`, `INSERT`, `UPDATE`, `DELETE`, and `CREATE`.

## SQL 2. Create a table and insert records

`CREATE TABLE` defines a table structure, and `INSERT INTO` adds records.

```python
import sqlite3
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()
cursor.execute("CREATE TABLE employees (employee_id INTEGER PRIMARY KEY, name TEXT NOT NULL, department TEXT, salary INTEGER)")
employees = [(1, "Anu", "IT", 50000), (2, "Ravi", "HR", 42000), (3, "Meena", "IT", 65000), (4, "John", "Sales", 45000), (5, "Sara", "HR", 52000)]
cursor.executemany("INSERT INTO employees VALUES (?, ?, ?, ?)", employees)
conn.commit()
print("Employees inserted:", cursor.execute("SELECT COUNT(*) FROM employees").fetchone()[0])
```

## SQL 3. What is SELECT?

`SELECT` retrieves data from a table. Use `*` for all columns or name the columns you need.

```python
cursor.execute("SELECT employee_id, name, department, salary FROM employees ORDER BY employee_id")
for row in cursor.fetchall():
    print(row)
```

## SQL 4. What is the WHERE clause?

`WHERE` filters rows that match a condition.

```python
cursor.execute("SELECT name, salary FROM employees WHERE salary > 48000 ORDER BY salary DESC")
print(cursor.fetchall())
```

## SQL 5. What is ORDER BY?

`ORDER BY` sorts results in ascending (`ASC`) or descending (`DESC`) order.

```python
cursor.execute("SELECT name, salary FROM employees ORDER BY salary DESC")
print(cursor.fetchall())
```

## SQL 6. What are aggregate functions?

`COUNT`, `SUM`, `AVG`, `MIN`, and `MAX` calculate values across multiple rows.

```python
cursor.execute("SELECT COUNT(*), AVG(salary), MIN(salary), MAX(salary) FROM employees")
count, average, minimum, maximum = cursor.fetchone()
print("Count:", count)
print("Average salary:", round(average, 2))
print("Minimum salary:", minimum)
print("Maximum salary:", maximum)
```

## SQL 7. What is GROUP BY?

`GROUP BY` groups rows with the same value, often to calculate an aggregate for each group.

```python
cursor.execute("SELECT department, COUNT(*) AS employee_count, AVG(salary) AS average_salary FROM employees GROUP BY department ORDER BY department")
for row in cursor.fetchall():
    print(row)
```

## SQL 8. Difference between WHERE and HAVING

- `WHERE` filters individual rows before grouping.
- `HAVING` filters grouped results after `GROUP BY`.

```python
cursor.execute("SELECT department, COUNT(*) FROM employees WHERE salary >= 45000 GROUP BY department HAVING COUNT(*) >= 1 ORDER BY department")
print(cursor.fetchall())
```

## SQL 9. What is a JOIN?

A `JOIN` combines related rows from tables. An `INNER JOIN` returns only rows that match the join condition.

```python
cursor.execute("CREATE TABLE departments (department TEXT PRIMARY KEY, manager TEXT)")
cursor.executemany("INSERT INTO departments VALUES (?, ?)", [("IT", "Priya"), ("HR", "Kiran"), ("Sales", "Arun")])
cursor.execute("SELECT e.name, e.department, d.manager FROM employees e INNER JOIN departments d ON e.department = d.department ORDER BY e.employee_id")
for row in cursor.fetchall():
    print(row)
```

## SQL 10. Primary key vs foreign key

- **Primary key:** uniquely identifies each row and cannot contain duplicate or `NULL` values.
- **Foreign key:** references a key in another table and helps maintain relationships between tables.

## SQL 11. What is a subquery?

A subquery is a query nested inside another SQL query.

```python
cursor.execute("SELECT name, salary FROM employees WHERE salary > (SELECT AVG(salary) FROM employees) ORDER BY salary DESC")
print(cursor.fetchall())
```

## SQL 12. DELETE vs TRUNCATE vs DROP

- `DELETE` removes selected rows and can use `WHERE`.
- `TRUNCATE` removes all rows while keeping the table structure in database systems that support it.
- `DROP` removes the table itself.
- SQLite does not support `TRUNCATE TABLE`; behavior varies by database.

## SQL 13. What is a database index?

An index can help a database find rows more efficiently. It uses storage and may slow down inserts or updates.

```python
cursor.execute("CREATE INDEX idx_employees_department ON employees(department)")
print("Index created for department.")
```

## SQL 14. How do UPDATE and DELETE work?

Use `UPDATE` to modify records and `DELETE` to remove records. Check the `WHERE` condition carefully.

```python
cursor.execute("INSERT INTO employees VALUES (?, ?, ?, ?)", (99, "Test User", "Intern", 20000))
cursor.execute("UPDATE employees SET salary = ? WHERE employee_id = ?", (22000, 99))
print("Updated:", cursor.execute("SELECT name, salary FROM employees WHERE employee_id = 99").fetchone())
cursor.execute("DELETE FROM employees WHERE employee_id = 99")
print("Remaining test rows:", cursor.execute("SELECT COUNT(*) FROM employees WHERE employee_id = 99").fetchone()[0])
```

## SQL 15. What is normalization?

Normalization organizes relational tables to reduce unnecessary duplication and improve data integrity. Common normal forms include 1NF, 2NF, and 3NF.

## SQL 16. UNION vs UNION ALL

- `UNION` combines query results and removes duplicates.
- `UNION ALL` combines query results and keeps duplicates.

```python
cursor.execute("SELECT department FROM employees WHERE department = 'IT' UNION SELECT department FROM employees WHERE department = 'HR' ORDER BY department")
print("UNION:", cursor.fetchall())
cursor.execute("SELECT department FROM employees WHERE department = 'IT' UNION ALL SELECT department FROM employees WHERE department = 'HR' ORDER BY department")
print("UNION ALL:", cursor.fetchall())
```

## SQL 17. Practice: Find the second-highest distinct salary

Use a subquery to find the largest salary below the maximum salary.

```python
cursor.execute("SELECT MAX(salary) FROM employees WHERE salary < (SELECT MAX(salary) FROM employees)")
print("Second-highest distinct salary:", cursor.fetchone()[0])
```

## SQL 18. What is NULL?

`NULL` represents missing or unknown data. Use `IS NULL` or `IS NOT NULL` to test it, not `= NULL`.

```python
cursor.execute("SELECT name FROM employees WHERE department IS NULL")
print(cursor.fetchall())  # No employees have a NULL department in this sample.
```
